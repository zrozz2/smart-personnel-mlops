```python
import joblib
from pathlib import Path
```

```python
# Load model

models_dir = Path("../models")

model_path = models_dir / "linear_regression_v1.pkl"

model = joblib.load(model_path)

print("Model berhasil dimuat!")
print("Tipe model:", type(model))
```

```python
# lihat struktur pipeline

print(model)
```

```python
# ambil preprocessing

preprocessor = model.named_steps["preprocessor"]

print("Preprocessor:")
print(preprocessor)
```

```python
# lihat nama feature input

feature_names = preprocessor.feature_names_in_

print("Jumlah feature input:", len(feature_names))

print("\nFeature yang digunakan model:")
for i, feature in enumerate(feature_names, start=1):
    print(f"{i}. {feature}")
```

```python
# periksa feature yang dibutuhkan

print("Jumlah feature:", len(feature_names))
print(feature_names.tolist())
```

```python
# melihat data asli

import pandas as pd

data_path = Path("../data/processed/employee_workload_feature_engineered.csv")

df = pd.read_csv(data_path)

print("Shape:", df.shape)

display(df.head())
```

```python
print("Kolom dataset:")
for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")
    
```
